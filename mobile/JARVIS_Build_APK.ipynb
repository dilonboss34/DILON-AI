```python
# هەنگاوی 2: دامەزراندنی Buildozer
!pip install -q --upgrade buildozer
!pip install -q --upgrade cython==0.29.37
print('✅ Buildozer installed!')
```

```python
# هەنگاوی 3: دانانی فایلەکانی پڕۆژەکە
# ئەم سەلولە فایلی zip ئەوێ کە دایلۆد کردوتە

import os
from google.colab import files

print('تکایە فایلی mobile.zip ئەپلۆد بکە...')
uploaded = files.upload()

# دەرهێنانی فایلەکان
zip_name = list(uploaded.keys())[0]
!unzip -o {zip_name} -d /content/jarvis_build/

# ئەگەر ناو فۆڵدەرێکدا بوو
import glob
spec_files = glob.glob('/content/jarvis_build/**/buildozer.spec', recursive=True)
if spec_files:
    build_dir = os.path.dirname(spec_files[0])
    print(f'✅ پڕۆژەکە دۆزرایەوە: {build_dir}')
else:
    build_dir = '/content/jarvis_build'
    print('❌ buildozer.spec نەدۆزرایەوە، تکایە دووبارە بکە')

print(f'Build directory: {build_dir}')
os.chdir(build_dir)
!ls -la
```

```python
# هەنگاوی 4: دانانی کۆدی API کەیت
# ئەمەیە API Key ئەوەی لە Gemini وەرگرتووتە

API_KEY = ''  # <-- API Key ی خۆت لێرە دابنێ، وەک: 'AIza...'

if API_KEY:
    # API Key دەخەینە ناو فایلێکی .env
    with open('.env', 'w') as f:
        f.write(f'GEMINI_API_KEY={API_KEY}\n')
    print('✅ API Key تۆمار کرا')
else:
    print('⚠️ API Key زیادت نەکراوە - ئەپەکە بەبێ هوش کار دەکات')
```

```python
# هەنگاوی 5: دروستکردنی APK  (ئەمە کاتی زیاتر دەبات - 15-25 خولەک)
import subprocess

print('🔨 دەستپێکردنی دروستکردنی APK...')
print('چاوەڕێ بکە - ئەمە 15-25 خولەک دەبات!')
print('=' * 50)

# دروستکردنی APK
result = subprocess.run(
    ['buildozer', '-v', 'android', 'debug'],
    capture_output=False,
    text=True
)

print('=' * 50)
if result.returncode == 0:
    print('✅ APK بە سەرکەوتوویی دروست کرا!')
else:
    print('❌ هەڵەیەک روویدا - تکایە هەڵەکانی سەرەوە بخوێنەوە')
```

```python
# هەنگاوی 6: داونلۆدکردنی APK
import glob
from google.colab import files

apk_files = glob.glob('bin/*.apk')

if apk_files:
    apk_path = apk_files[0]
    print(f'✅ APK دۆزرایەوە: {apk_path}')
    print('داونلۆد دەکرێت...')
    files.download(apk_path)
    print('✅ داونلۆد دەستپێکرد!')
    print('')
    print('📱 ئێستا فایلی APK بۆ مۆبایلەکەت بنێرە و ئینستاڵی بکە!')
else:
    print('❌ APK نەدۆزرایەوە، دەبێت هەڵەکانی هەنگاوی 5 بخوێنەوە')
    !ls -la bin/ 2>/dev/null || echo 'فۆڵدەری bin نەدۆزرایەوە'
```
